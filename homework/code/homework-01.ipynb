#### Q4. Missing Values
How many columns in the dataset have missing values?

```python
df.info()
```

```python
# Function to display information about missing value
def show_null(df):
    """
    Return the total missing values and the percentage of
    missing values by column.
    """
    null_count = df.isnull().sum()
    null_percentage = (null_count / df.shape[0]) * 100
    empty_count = pd.Series(((df == ' ') | (df == '')).sum())
    empty_percentage = (empty_count / df.shape[0]) * 100
    nan_count = pd.Series(((df == 'nan') | (df == 'NaN') | (df == np.nan)).sum())
    nan_percentage = (nan_count / df.shape[0]) * 100
    return pd.DataFrame({'num_missing': null_count,
                         'missing_percentage': null_percentage,
                         'num_empty': empty_count,
                         'empty_percentage': empty_percentage,
                         'nan_count': nan_count,
                         'nan_percentage': nan_percentage})

show_null(df)
```

```python
missing_cols = (df.isnull().sum() > 0).sum()
print("Columns with missing values:", missing_cols)
```

#### Q6. Median Value of Horsepower
1. Find the median value of the horsepower column in the dataset.
2. Next, calculate the most frequent value of the same horsepower column.  
3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.  
4. Now, calculate the median value of horsepower once again.  

Has it changed?

```python
# Check the initial data descriptive statistics
df.describe()
```

```python
# HP initial median and mode
hp_median_before = df['horsepower'].median()
hp_mode = df['horsepower'].mode().iloc[0]

# Fill missing values with mode
df_imputed = df.copy()
df_imputed['horsepower'] = df_imputed['horsepower'].fillna(hp_mode)

hp_median_after = df_imputed['horsepower'].median()

print(f"HP Before: {hp_median_before}, HP Mode: {hp_mode}, HP After: {hp_median_after}")
```

#### Q7. Sum of Weights (Linear Regression Normal Equation)
1. Select all the cars from Asia
2. Select only columns `vehicle_weight` and `model_year`
3. Select the first 7 values
4. Get the underlying NumPy array. Let's call it `X`.
5. Compute matrix-matrix multiplication between the transpose of `X` and `X`. To get the transpose, use `X.T`. Let's call the result `XTX`.
6. Invert `XTX`.
7. Create an array `y` with values `[1100, 1300, 800, 900, 1000, 1100, 1200]`.
8. Multiply the inverse of `XTX` with the transpose of `X`, and then multiply the result by `y`. Call the result `w`.
9. What's the sum of all the elements of the result?

```python
asia_subset = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = asia_subset.to_numpy()

# Computation
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

print(f"Weights w: {w}")
print(f"Sum of weights: {w.sum():.3f}")
```
